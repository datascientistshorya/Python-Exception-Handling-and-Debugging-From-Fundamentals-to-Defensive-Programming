#### Q1. Build a Robust Order Processor
Write a function called process_order(data) that processes an order dictionary and handles these exceptions:
- KeyError — a required key is missing.
- ValueError — a value cannot be converted to the required type.
- TypeError — a value has an unexpected data type.

orders = [
    {"product": "Laptop", "quantity": "2", "price": "50000"},
    {"product": "Phone", "quantity": "abc", "price": "20000"},
    {"quantity": "3", "price": "15000"},
    {"product": "Tablet", "quantity": None, "price": "10000"}
]

Requirements:
1. Extract product, quantity, and price from each order.
2. Convert quantity and price into numeric values.
3. Calculate total = quantity * price.
4. Handle each exception separately with an appropriate message.
5. Ensure one invalid order does not prevent the remaining orders from being processed.
6. Return or print the result for each valid order.


```python

orders = [
    {"product": "Laptop", "quantity": "2", "price": "50000"},
    {"product": "Phone", "quantity": "abc", "price": "20000"},
    {"quantity": "3", "price": "15000"},
    {"product": "Tablet", "quantity": None, "price": "10000"}
]

def process_order(data):
    results = []

    for order in data:
        try:
            product = order["product"]
            quantity = float(order["quantity"])
            price = float(order["price"])

            total = quantity * price

            result = {
                "product": product,
                "quantity": quantity,
                "price": price,
                "total": total
            }

            results.append(result)
            print(f"Order processed: {product}, Total = {total}")

        except KeyError as e:
            print(f"KeyError: Missing required key {e}")

        except ValueError as e:
            print(f"ValueError: Invalid numeric value - {e}")


        except TypeError as e:
            print(f"TypeError: Unexpected data type - {e}")

    return results

valid_orders = process_order(orders)
print("\nValid orders:", valid_orders)
```

#### Q8. Validate Dictionary Data
Write validate_customer(customer) that checks:
customer = {    "name": "Aman",    "age": 25,    "email": "aman@example.com"}



Requirements:
- Handle missing keys.
- Validate that the name is nonempty.
- Validate that age is a reasonable positive integer.
- Raise meaningful exceptions for invalid data.

```python

customer = {
    "name": "Aman",
    "age": 25,
    "email": "aman@example.com"
}

def validate_customer(customer):
    try:
        name = customer["name"]
        age = customer["age"]
        email = customer["email"]

        if not isinstance(name, str) or not name.strip():
            raise ValueError("Customer name must be a nonempty string.")

        if isinstance(age, bool) or not isinstance(age, int):
            raise TypeError("Customer age must be an integer.")

        if age <= 0 or age > 120:
            raise ValueError("Customer age must be between 1 and 120.")

        if not isinstance(email, str) or not email.strip():
            raise ValueError("Customer email must be a nonempty string.")

        return "Customer data is valid."

    except KeyError as e:
        return f"Missing required field: {e}"

    except (ValueError, TypeError) as e:
        return f"Invalid customer data: {e}"


print(validate_customer(customer))

customer["age"] = 250
print(validate_customer(customer))

customer.pop("email")
print(validate_customer(customer))
```

#### Q9. Build a Retry Mechanism
Write a function that asks the user to enter an integer, allowing a maximum of three attempts.
Requirements:
- Handle invalid input using ValueError.
- Stop immediately when valid input is received.
- Print an appropriate message if all attempts fail.

```python

attempt = 0
chance = 3

while attempt < 3:
    try:
        entry = input("Enter an integer: ")
        attempt += 1

        number = int(entry)

        print("You entered:", number)
        break

    except ValueError:
        chance -= 1
        print("Invalid entry. Please enter a valid integer.")

        if chance > 0:
            print(chance, "tries remaining")
        else:
            print("No tries remaining.")
```

#### Q10. Mini-Project — Robust Order Processor
Build a complete order-processing program using:

orders = [    

{"product": "Laptop", "quantity": 2, "price": 50000}, 

{"product": "Phone", "quantity": 3, "price": 20000},   

{"product": "Tablet", "quantity": 0, "price": 15000}, 

{"product": "Monitor", "quantity": "two", "price": 12000},  

{"quantity": 4, "price": 5000},  

{"product": "Keyboard", "quantity": 5, "price": 1500}, 

{"product": "Mouse", "quantity": 2, "price": -500}]



Requirements:
1. Validate all required fields.
2. Convert valid numeric strings when appropriate.
3. Reject invalid types, non-positive quantities, and negative prices.
4. Handle exceptions with specific error messages.
5. Continue processing after individual failures.
6. Store valid and rejected orders separately.
7. Calculate revenue for each valid product.
8. Calculate total revenue.
9. Report the number of valid and rejected orders.
10. Keep the code modular and readable.

```python

orders = [
    {"product": "Laptop", "quantity": 2, "price": 50000},
    {"product": "Phone", "quantity": 3, "price": 20000},
    {"product": "Tablet", "quantity": 0, "price": 15000},
    {"product": "Monitor", "quantity": "two", "price": 12000},
    {"quantity": 4, "price": 5000},
    {"product": "Keyboard", "quantity": 5, "price": 1500},
    {"product": "Mouse", "quantity": 2, "price": -500}
]

def validate(orders):
    clean = []
    err = 0
    valid = 0

    for order in orders:
        try:
            prod = order["product"]
            quan = float(order["quantity"])
            price = float(order["price"])

            if quan <= 0 or price <= 0:
                raise ValueError("Quantity and price must be positive.")

            revenue = quan * price

            clean.append({
                "product": prod,
                "quantity": quan,
                "price": price,
                "revenue": revenue
            })

            valid += 1
            print(f"{prod}: revenue = {revenue:.2f}")

        except KeyError as e:
            print("Missing required key:", e)
            err += 1

        except ValueError as e:
            print("Invalid numeric value:", e)
            err += 1

        except TypeError as e:
            print("Unexpected data type:", e)
            err += 1

    print("Total valid orders:", valid)
    print("Total invalid orders:", err)

    return clean


clean = validate(orders)
print(clean)
total = sum(order["revenue"] for order in clean)
print("Total revenue:", total)
```
