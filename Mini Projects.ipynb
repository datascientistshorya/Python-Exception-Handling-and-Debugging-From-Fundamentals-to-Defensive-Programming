#### Q2. Banking Transaction System
Create a simple banking system with deposit and withdrawal operations.
account = {    "account_holder": "Aman",    "balance": 10000}transactions = [    {"type": "deposit", "amount": 2000},    {"type": "withdraw", "amount": 3000},    {"type": "withdraw", "amount": 15000},    {"type": "deposit", "amount": -500},    {"type": "transfer", "amount": 1000},    {"type": "withdraw", "amount": "abc"}]



Requirements:
- Create a custom InsufficientFundsError.
- Reject zero or negative transaction amounts.
- Handle invalid transaction types.
- Validate numeric amounts.
- Prevent withdrawals exceeding the balance.
- Process transactions independently.
- Maintain the correct balance and generate a transaction report.

```python

account = {"account_holder": "Aman", "balance": 10000}

transactions = [
    {"type": "deposit", "amount": 2000},
    {"type": "withdraw", "amount": 3000},
    {"type": "withdraw", "amount": 15000},
    {"type": "deposit", "amount": -500},
    {"type": "transfer", "amount": 1000},
    {"type": "withdraw", "amount": "abc"}
]

def validate(transactions):
    clean = []

    for transaction in transactions:
        try:
            type_ = transaction["type"]
            amount = int(transaction["amount"])

            if amount <= 0:
                raise ValueError("Amount must be positive.")

            if type_ == "deposit":
                account["balance"] += amount

            elif type_ in ("withdraw", "transfer"):
                if amount > account["balance"]:
                    raise ValueError("Insufficient balance.")

                account["balance"] -= amount

            else:
                raise ValueError("Unsupported transaction type.")

            transaction["balance"] = account["balance"]
            clean.append(transaction)

            print(
                f"{type_.title()} successful. "
                f"Remaining balance: {account['balance']}"
            )

        except KeyError as e:
            print("Missing transaction field:", e)

        except (ValueError, TypeError) as e:
            print("Invalid transaction:", e)

    return clean


valid_transactions = validate(transactions)

print("\nValid transactions:", valid_transactions)
print("Final account balance:", account["balance"])
```

#### Q3. Expense Tracker
Build an expense tracker that validates and summarizes personal expenses.
expenses = [    {"category": "Food", "amount": 250},    {"category": "Travel", "amount": "500"},    {"category": "Shopping", "amount": -300},    {"category": "Food", "amount": 150},    {"category": "", "amount": 200},    {"category": "Bills", "amount": None},    {"amount": 100}]



Requirements:
- Validate category and amount.
- Convert valid numeric strings.
- Reject negative amounts, missing fields, and empty categories.
- Continue processing valid entries.
- Calculate total spending by category.
- Find the highest-spending category.
- Report valid and rejected records with reasons.

```python

expenses = [
    {"category": "Food", "amount": 250},
    {"category": "Travel", "amount": "500"},
    {"category": "Shopping", "amount": -300},
    {"category": "Food", "amount": 150},
    {"category": "", "amount": 200},
    {"category": "Bills", "amount": None},
    {"amount": 100}
]

def validate(expenses):
    clean = []
    invalid = []
    total = 0

    for expense in expenses:
        try:
            cat = expense["category"]
            amount = int(expense["amount"])

            if not isinstance(cat, str) or not cat.strip():
                raise ValueError("Category cannot be empty.")

            if amount < 0:
                raise ValueError("Amount cannot be negative.")

            total += amount
            clean.append(expense)

        except KeyError as e:
            invalid.append(expense)
            print("Missing required key:", e)

        except ValueError as e:
            invalid.append(expense)
            print("Invalid value:", e)

        except TypeError as e:
            invalid.append(expense)
            print("Unexpected data type:", e)

    return clean, invalid, total

def spending_by_category(clean):
    category_totals = {}

    for expense in clean:
        category = expense["category"]
        amount = int(expense["amount"])

        category_totals[category] = (
            category_totals.get(category, 0) + amount
        )

    return category_totals

def highest_spending_category(category_totals):
    if not category_totals:
        return None

    return max(category_totals, key=category_totals.get)

clean, invalid, total = validate(expenses)
category_spending=spending_by_category(clean)

print("\nClean transactions:", clean)
print("Invalid transactions:", invalid)
print("Total expenses:", total)
print('spending by category', category_spending)
print('highest spend on ',highest_spending_category(category_spending))
```
