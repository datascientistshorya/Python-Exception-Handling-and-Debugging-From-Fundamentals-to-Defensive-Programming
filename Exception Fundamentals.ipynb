#### Q2. Calculate Average with Error Handling
Create calculate_average(numbers) that:
- Returns the average of a list of numbers.
- Returns "Cannot calculate average of an empty list" for an empty list.
- Returns "Invalid input" if the list contains an invalid value that prevents numeric calculation.
- Uses try-except appropriately.

```python
def calculate_average(numbers):
    if len(numbers)==0:
        return 'Cannot calculate average of an empty list'

    try:
        total=0
        for i in numbers:
            if (isinstance(i,(int,float))):
                total+=i
         
        return ('avg: ',total/len(numbers))

    except (ValueError, TypeError):
        return "Invalid input"

print(calculate_average([10, 20, 30]))
print(calculate_average([]))
print(calculate_average([10, "hello", 30]))
    
            
```

#### Q3. Handle Multiple Exceptions
Write a function safe_divide(a, b) that:
- Returns a / b when valid.
- Returns "Cannot divide by zero" for division by zero.
- Returns "Invalid numbers" when the inputs cannot be used for division.

```python
def safe_divide(a, b):
    try:
        return a / b

    except ZeroDivisionError:
        return "Cannot divide by zero"

    except TypeError:
        return "Invalid numbers"


print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "hello"))
print(safe_divide(3.410,5.67))
```

## Q4. Use try-except-else
Write a function convert_to_integer(value).
Requirements:
- Attempt to convert value into an integer.
- If conversion fails, return "Conversion failed".
- If conversion succeeds, use the else block to return the converted integer.

```python
def convert_to_integer(value):
    try:
        number=int(value)
    except ValueError:
        return 'Invalid input,Conversion Failed'
    else:
        return number

print(convert_to_integer("125"))
print(convert_to_integer("hello"))
print(convert_to_integer("42"))
```

#### Q5. Understand the finally Block
Write a function process_data(value) that attempts to convert a value into a float.
Requirements:
- Return the converted float if successful.
- Return "Invalid number" if conversion raises ValueError or TypeError.
- Use finally to print "Processing completed" regardless of success or failure.

```python
def process_data(value):
    try:
        num=float(value)
    except (ValueError,TypeError):
        return 'Invalid Input'
    else:
        return num
    finally:
        print('processing Complete')

print(process_data("25.5"))
print(process_data("hello"))
print(process_data(None))
```

#### Q6. Raise Your Own Exception
Create a function calculate_discount(price, discount_percent).
Requirements:
- Raise a ValueError if price is negative.
- Raise a ValueError if the discount percentage is below 0 or above 100.
- Otherwise, return the final price after applying the discount.

```python
def  calculate_discount(price, discount_percent):
    try:
        if price<0 or not (0<=discount_percent<=100):
            raise ValueError
        final_price=price-(discount_percent/100)*price
    except (ValueError):
        return 'Invalid input'
    else:
        return final_price

print(calculate_discount(1000, 20))
print(calculate_discount(-100, 20))
print(calculate_discount(1000, 120))
```

#### Q7. Build a Data-Cleaning Function
A dataset contains numeric strings, missing values, and invalid text.

Write clean_numeric_data(data) that:
- Converts valid numeric values to floats.
- Skips None, empty strings, and invalid numeric strings.
- Returns a list containing only valid numbers.
- Uses exception handling where appropriate.

```python
def clean_numeric_data(data):
    clean=[]
    for i in data:
        try:
            num=float(i)
        except (ValueError,TypeError):
            print(' list Item not valid')
        else:
            clean.append(num)
    return ('cleaned list', clean)

raw_data = ["100", "250", None, "unknown", "400", "", "150"]
print(clean_numeric_data(raw_data))
            
            
```

#### Q8. Validate an Order Before Processing
Write process_order(order) that validates a dictionary containing product, quantity, and price.
Example:

order = {    "product": "Laptop",    "quantity": 2,    "price": 50000}



Requirements:
- Raise ValueError if quantity is zero or negative.
- Raise ValueError if price is negative.
- Handle a missing required key using KeyError.
- Return the total order value if valid.
- Use try-except to return meaningful error messages.

```python
def process_order(order):
    try:
        quantity = order['quantity']
        price = order['price']

        if quantity <= 0:
            raise ValueError("Quantity must be greater than zero")

        if price <= 0:
            raise ValueError("Price must be greater than zero")

        return ('Total order value: ', quantity * price)

    except (ValueError, TypeError):
        return "Invalid quantity or price"

    except KeyError:
        return "Missing order information"


order = { "product": "Laptop", "quantity": 2, "price": 50000}
print(process_order(order))
order = { "product": "Laptop", "quantity": -2, "price": 50000}
print(process_order(order))
order = { "product": "Laptop", "quantity": 24, "price": -50000}
print(process_order(order))

order = { "product": "Laptop", "": -2, "price": 50000}
print(process_order(order))
```

#### Q9. Debug a Sales Analytics Function
The following function contains multiple problems.

def calculate_revenue(sales):
    total = 0

    for sale in sales:
        total += float(sale["amount"])

    return total


sales = [
    {"amount": "1000"},
    {"amount": "2500"},
    {"amount": "unknown"},
    {"amount": "1500"},
    {"price": "500"}
]

print(calculate_revenue(sales))

Your tasks:
1. Identify the exceptions that can occur.
2. Rewrite the function so invalid records do not stop the entire calculation.
3. Count the valid and rejected records.
4. Return the total revenue, valid record count, and rejected record count.

```python

def calculate_revenue(sales):
    total = 0
    valid_count = 0
    rejected_count = 0

    for sale in sales:
        try:
            amount = float(sale["amount"])

        except (ValueError, TypeError, KeyError):
            rejected_count += 1

        else:
            total += amount
            valid_count += 1

    return total, valid_count, rejected_count


sales = [
    {"amount": "1000"},
    {"amount": "2500"},
    {"amount": "unknown"},
    {"amount": "1500"},
    {"price": "500"}
]

total, valid, rejected = calculate_revenue(sales)

print("Total revenue:", total)
print("Valid records:", valid)
print("Rejected records:", rejected)
```

#### Q10. Mini-Project — Robust Sales Data Processor
Build a small sales data processor that combines the skills from this batch.
Input dataset:

sales_data = [
    {"product": "Laptop", "quantity": 2, "price": 50000},
    {"product": "Phone", "quantity": 3, "price": 20000},
    {"product": "Tablet", "quantity": 0, "price": 15000},
    {"product": "Monitor", "quantity": 2, "price": -5000},
    {"product": "Keyboard", "quantity": "unknown", "price": 2000},
    {"product": "Mouse", "quantity": 4, "price": 500},
    {"product": "Headphones", "quantity": 2}
]




Create a function analyze_sales(sales_data).
Your program must:
1. Validate that each record contains product, quantity, and price.
2. Reject missing fields, invalid numeric values, non-positive quantities, and negative prices.
3. Use custom ValueError messages for invalid business values.
4. Use try-except to handle bad records without terminating the entire analysis.
5. Calculate total revenue from valid records only.
6. Count valid and rejected records.
7. Track revenue by product.
8. Use else for successfully processed records.
9. Use finally to print "Sales analysis completed" after the analysis.
10. Return a dictionary containing the results.

```python

def analyze_sales(sales_data):
    total_revenue = 0
    valid_count = 0
    rejected_count = 0
    revenue_by_product = {}

    try:
        for sale in sales_data:
            try:
                # 1. Validate required fields
                if not all(key in sale for key in
                           ("product", "quantity", "price")):
                    raise ValueError("Missing required fields")

                product = sale["product"]

                # 2. Convert numeric values
                quantity = float(sale["quantity"])
                price = float(sale["price"])

                # 3. Validate business rules
                if quantity <= 0:
                    raise ValueError(
                        "Quantity must be greater than zero"
                    )

                if price < 0:
                    raise ValueError(
                        "Price cannot be negative"
                    )

                # 4. Calculate revenue
                revenue = quantity * price

            except (ValueError, TypeError, KeyError) as error:
                rejected_count += 1
                print(f"Rejected record: {error}")

            else:
                # Runs only for valid records
                total_revenue += revenue
                valid_count += 1

                revenue_by_product[product] = revenue

    finally:
        print("Sales analysis completed")

    return {
        "total_revenue": total_revenue,
        "valid_records": valid_count,
        "rejected_records": rejected_count,
        "revenue_by_product": revenue_by_product
    }


sales_data = [
    {"product": "Laptop", "quantity": 2, "price": 50000},
    {"product": "Phone", "quantity": 3, "price": 20000},
    {"product": "Tablet", "quantity": 0, "price": 15000},
    {"product": "Monitor", "quantity": 2, "price": -5000},
    {"product": "Keyboard", "quantity": "unknown", "price": 2000},
    {"product": "Mouse", "quantity": 4, "price": 500},
    {"product": "Headphones", "quantity": 2}
]

result = analyze_sales(sales_data)
print(result)
```
